```python
%matplotlib inline
import pickle
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

plt.rcParams.update({'font.size': 10})

with open("../data/Dev_swap_results/plot_data.pkl", "rb") as f:
    plot_data = pickle.load(f)

categories = ['Endogenous', 'High Activity Background', 'Mid Activity Background', 'Low Activity Background', 'Dishuffled_background']
xtick_labels = ['Endogenous', 'High Activity\nBackground', 'Mid Activity\nBackground', 'Low Activity\nBackground', 'Dinucleotide-shuffled\nbackground']
source_bins = ['high', 'mid', 'low']
colors = ['#D32F2F', '#1976D2', '#388E3C']
width = 0.25
n_bins = len(source_bins)

fig, ax = plt.subplots(figsize=(10, 6), dpi=300)
positions = [[i * (n_bins * width + 0.4) + j * width for j in range(n_bins)] for i in range(len(categories))]

for i, cat in enumerate(categories):
    data_for_cat = [plot_data[b][cat] for b in source_bins]
    bp = ax.boxplot(data_for_cat, positions=positions[i], widths=width*0.8, patch_artist=True, showfliers=False,
                    medianprops={'color': 'black'})
    for patch, color in zip(bp['boxes'], colors):
        patch.set_facecolor(color); patch.set_alpha(0.7)
        if cat == 'Endogenous':
            patch.set_hatch('...'); patch.set_edgecolor('black'); patch.set_linewidth(1.5)
        elif cat == 'Dishuffled_background':
            patch.set_hatch('xxx'); patch.set_edgecolor('black'); patch.set_linewidth(1.5)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_xticks([np.mean(p) for p in positions])
ax.set_xticklabels(xtick_labels, rotation=0, ha='center')
ax.set_ylim(-3, 7)
ax.legend(handles=[Patch(facecolor=colors[i], alpha=0.7, label=f'{b.capitalize()} source') for i, b in enumerate(source_bins)],
          loc='upper left', bbox_to_anchor=(1.02, 1.0), borderaxespad=0., frameon=False)
ax.set_ylabel("Dev Prediction Score")
ax.axhline(y=0, color='black', linestyle='-', alpha=0.5)
plt.tight_layout()
plt.savefig("final_Motif_Swap_Dev_results.pdf", dpi=300, bbox_inches='tight')
plt.show()
```
