```python
# Load hits for all three bins
hits_high = pd.read_csv("../foreground_analysis/TF-Modisco-lite_results/Dev_manual_hits/high_clean_hits.csv")
hits_mid = pd.read_csv("../foreground_analysis/TF-Modisco-lite_results/Dev_manual_hits/mid_clean_hits.csv")
hits_low = pd.read_csv("../foreground_analysis/TF-Modisco-lite_results/Dev_manual_hits/low_clean_hits.csv")

# Get valid sequences for each bin
valid_seqs_high = hits_high["sequence_name"].unique()
valid_seqs_mid = hits_mid["sequence_name"].unique()
valid_seqs_low = hits_low["sequence_name"].unique()

# Group by sequence for each bin
hits_byseq = {
    'high': dict(tuple(hits_high.groupby("sequence_name"))),
    'mid': dict(tuple(hits_mid.groupby("sequence_name"))),
    'low': dict(tuple(hits_low.groupby("sequence_name")))
}

# Store valid sequences
valid_seqs = {
    'high': valid_seqs_high,
    'mid': valid_seqs_mid,
    'low': valid_seqs_low
}

```

```python
import re
import ast
import torch
import numpy as np

# Remove trailing dtype/device hints like ", dtype=torch.int8)" or ", device='cuda:0')"
_DTYPE_DEVICE_RE = re.compile(r",(?:\s*dtype\s*=\s*[\w\.]+|\s*device\s*=\s*[^)\]]+)\s*\)?\s*$")

def _extract_bracket_payload(s: str) -> str:
    s = s.strip()
    # If starts with tensor( ... ), drop the leading "tensor(" and trailing ")"
    if s.startswith("tensor(") and s.endswith(")"):
        s = s[len("tensor("):-1].strip()
    # Keep only the bracketed list from the first '[' to its matching ']'
    start = s.find('[')
    if start < 0:
        raise ValueError("No bracketed payload in string")
    # Remove dtype/device suffixes after the closing bracket region
    s = _DTYPE_DEVICE_RE.sub("", s)
    # Now trim to balanced brackets
    depth = 0
    end = None
    for i, ch in enumerate(s[start:], start=start):
        if ch == '[':
            depth += 1
        elif ch == ']':
            depth -= 1
            if depth == 0:
                end = i + 1
                break
    if end is None:
        raise ValueError("Unbalanced brackets in payload")
    return s[start:end]

def str_tensor_to_torch(s: str, dtype=None, device=None):
    if s is None:
        return None
    payload = _extract_bracket_payload(str(s))
    data = ast.literal_eval(payload)  # only literal content
    use_dtype = dtype if dtype is not None else torch.float32
    return torch.tensor(data, dtype=use_dtype, device=device)

COL_DTYPES = {
    "Cluster_foreground": torch.float32,
    "Scaled_foreground": torch.float32,
    "Avg_background": torch.float32,
    "ohe_seq": torch.int8,
}

def to_torch(v, dtype, device=None):
    if v is None:
        return None
    if isinstance(v, str):
        return str_tensor_to_torch(v, dtype=dtype, device=device)
    if isinstance(v, torch.Tensor):
        return v.to(dtype=dtype, device=device) if device else v.to(dtype=dtype)
    return torch.tensor(v, dtype=dtype, device=device)

def restore_tensor_columns(df, cols, dtypes, device=None):
    for col in cols:
        if col in df.columns:
            dt = dtypes.get(col, torch.float32)
            df[col] = [to_torch(v, dtype=dt, device=device) for v in df[col].tolist()]
    return df

# Load all three bins
high_df = pd.read_csv("../experimental_library_generation/Binned_libraries/Dev_full_data/high_df.csv")
mid_df = pd.read_csv("../experimental_library_generation/Binned_libraries/Dev_full_data/mid_df.csv")
low_df = pd.read_csv("../experimental_library_generation/Binned_libraries/Dev_full_data/low_df.csv")

high_df = restore_tensor_columns(
    high_df,
    cols=["Cluster_foreground", "Scaled_foreground", "Avg_background", "ohe_seq"],
    dtypes=COL_DTYPES,
)
mid_df = restore_tensor_columns(
    mid_df,
    cols=["Cluster_foreground", "Scaled_foreground", "Avg_background", "ohe_seq"],
    dtypes=COL_DTYPES,
)
low_df = restore_tensor_columns(
    low_df,
    cols=["Cluster_foreground", "Scaled_foreground", "Avg_background", "ohe_seq"],
    dtypes=COL_DTYPES,
)

high_ohe_list = high_df["ohe_seq"].values
high_att_list = high_df["Scaled_foreground"].values
mid_ohe_list = mid_df["ohe_seq"].values
mid_att_list = mid_df["Scaled_foreground"].values
low_ohe_list = low_df["ohe_seq"].values
low_att_list = low_df["Scaled_foreground"].values

# Create a combined lookup
bin_data = {
    'high': {'ohe': high_ohe_list, 'att': high_att_list, 'df': high_df},
    'mid': {'ohe': mid_ohe_list, 'att': mid_att_list, 'df': mid_df},
    'low': {'ohe': low_ohe_list, 'att': low_att_list, 'df': low_df}
}
```

```python
def get_spacings_from_lengths(regions):
    starts = regions['start'].to_numpy()
    ends = regions['end'].to_numpy()
    lengths = (ends - starts)
    spacings = (starts[1:] - ends[:-1]).astype(int).tolist() if len(starts) > 1 else []
    return spacings
```

```python
# make targets for all bins
# Targets = original sequence with shuffled motifs substituted in
# This preserves the background but randomizes the motifs
n_seqs = 100
targets_byseq = {}  # targets_byseq[bin][seq_idx] = list of 10 shuffled backgrounds
print("Making all targets for all bins")

for bin_name in ['high', 'mid', 'low']:
    targets_byseq[bin_name] = {}
    ohe_list = bin_data[bin_name]['ohe']
    first = valid_seqs[bin_name][0:n_seqs]  # Bin-specific sequences
    
    print(f"\nProcessing {bin_name} bin...")
    for idx, v in enumerate(first):
        ohe = ohe_list[v]  # (4, L) - original sequence
        ohe_batch = ohe.unsqueeze(0)  # (1, 4, L)
        seq_len = ohe.shape[-1]
        
        # Get motif positions for this bin
        regions = hits_byseq[bin_name][v][['start','end']].sort_values(by='start',ascending=True).reset_index(drop=True)
        starts = regions['start'].to_numpy()
        ends = regions['end'].to_numpy()
        spacings = get_spacings_from_lengths(regions)
        
        # Debug: print info about negative spacings
        if len(spacings) > 0 and min(spacings) < 0:
            print(f"\nDEBUG: {bin_name} bin, seq {v} (idx {idx}):")
            print(f"  Seq len: {seq_len}")
            print(f"  Motif positions: {list(zip(starts, ends))}")
            print(f"  Spacings: {spacings}")
        
        # Dinucleotide shuffle the entire sequence
        shuffles = dinucleotide_shuffle(ohe_batch, n=10)
        
        # Handle different possible output shapes from dinucleotide_shuffle
        if shuffles.ndim == 4:
            if shuffles.shape[0] == 10:  # (10, 1, 4, L)
                shuffles = shuffles.squeeze(1)  # (10, 4, L)
            elif shuffles.shape[1] == 10:  # (1, 10, 4, L)
                shuffles = shuffles.squeeze(0)  # (10, 4, L)
        
        targets = []
        for i in range(shuffles.shape[0]):
            shuffled_seq = shuffles[i]  # (4, L)
            # Extract motifs from the SHUFFLED sequence at the original positions
            motifs = [shuffled_seq[:, int(starts[j]):int(ends[j])].unsqueeze(0) for j in range(len(starts))]
            # Substitute these shuffled motifs into the ORIGINAL sequence
            subed_seq = multisubstitute(ohe_batch, motifs, spacing=spacings, start=int(starts[0]))
            targets.append(subed_seq)
        targets_byseq[bin_name][int(v)] = targets

print(f"\nTotal targets: {sum(len(targets_byseq[b]) for b in ['high', 'mid', 'low'])}")
print("Finished making all targets")
```

```python
# make sources for all bins
n_seqs = 100
sources_byseq = {}  # sources_byseq[bin][seq_idx] = list of motif tensors
print("Making all sources for all bins")

for bin_name in ['high', 'mid', 'low']:
    sources_byseq[bin_name] = {}
    ohe_list = bin_data[bin_name]['ohe']
    first = valid_seqs[bin_name][0:n_seqs]  # Bin-specific sequences
    
    print(f"\nProcessing {bin_name} bin...")
    for idx, v in enumerate(first):
        ohe = ohe_list[v].T
        regions = hits_byseq[bin_name][v][['start','end']].sort_values(by='start',ascending=True).reset_index(drop=True)

        starts = regions['start'].to_numpy()
        ends = regions['end'].to_numpy()
        
        # Debug: print info about negative spacings
        spacings_debug = get_spacings_from_lengths(regions)
        if len(spacings_debug) > 0 and min(spacings_debug) < 0:
            print(f"\nDEBUG: {bin_name} bin, seq {v} (idx {idx}):")
            print(f"  Motif positions: {list(zip(starts, ends))}")
            print(f"  Spacings: {spacings_debug}")
        
        motifs = [ohe[starts[j]:ends[j]].T.unsqueeze(0) for j in range(len(starts))]
        
        sources_byseq[bin_name][int(v)] = motifs

print(f"\nTotal sources: {sum(len(sources_byseq[b]) for b in ['high', 'mid', 'low'])}")
print("Finished making all sources")
```

```python
## All vs all sources vs targets across all bins
n_seqs = 100
# swaps[source_bin][source_seq][target_bin][target_seq] = list of 10 swapped seqs
swaps = {}
print("Making all-to-all swaps across bins")

for source_bin in ['high', 'mid', 'low']:
    swaps[source_bin] = {}
    first_source = valid_seqs[source_bin][0:n_seqs]  # Bin-specific source sequences
    
    print(f"\nProcessing {source_bin} source bin...")
    for v in first_source:
        motifs = sources_byseq[source_bin][v]
        regions = hits_byseq[source_bin][v][['start','end']].sort_values(by='start', ascending=True).reset_index(drop=True)
        spacings = get_spacings_from_lengths(regions)
        starts = regions['start'].to_numpy()
        
        # Debug: print info about negative spacings
        if len(spacings) > 0 and min(spacings) < 0:
            print(f"  DEBUG: {source_bin} source, seq {v}: spacings={spacings}")
        
        swaps[source_bin][v] = {}
        
        for target_bin in ['high', 'mid', 'low']:
            swaps[source_bin][v][target_bin] = {}
            first_target = valid_seqs[target_bin][0:n_seqs]  # Bin-specific target sequences
            
            for vt in first_target:
                if vt == v and source_bin == target_bin:
                    continue  # Skip self-swaps within same bin
                
                # Skip if target wasn't created (had overlapping motifs)
                if int(vt) not in targets_byseq[target_bin]:
                    continue
                    
                targets = targets_byseq[target_bin][vt]
                new_seqs = []
                for j in range(len(targets)):
                    bg = targets[j]
                    new_seq = multisubstitute(bg, motifs, spacing=spacings, start=int(starts[0]))
                    new_seqs.append(new_seq)
                swaps[source_bin][v][target_bin][vt] = new_seqs

print("\nDone making swaps.")

```

```python
# Make dishuffled backgrounds for all bins
n_seqs = 100
shuffledbgs_byseq = {}  # shuffledbgs_byseq[bin][seq_idx] = tensor of shape (10, C, L)
print("Making all dishuffled backgrounds for all bins")

for bin_name in ['high', 'mid', 'low']:
    shuffledbgs_byseq[bin_name] = {}
    ohe_list = bin_data[bin_name]['ohe']
    first = valid_seqs[bin_name][0:n_seqs]  # Bin-specific sequences
    
    print(f"\nProcessing {bin_name} bin...")
    for v in first:
        ohe = ohe_list[v].unsqueeze(0)
        regions = hits_byseq[bin_name][v][['start','end']].sort_values('start').reset_index(drop=True)
        spacings = get_spacings_from_lengths(regions)  # uses start[i+1] - end[i]
        starts = regions['start'].to_numpy()
        ends = regions['end'].to_numpy()
        
        # Debug: print info about negative spacings
        if len(spacings) > 0 and min(spacings) < 0:
            print(f"  DEBUG: {bin_name} bin, seq {v}: spacings={spacings}")

        # Generate 10 dinucleotide-shuffled backgrounds per sequence
        bgs = dinucleotide_shuffle(ohe, n=10).squeeze(0)  # shape: (10, C, L)
        ohe = ohe.squeeze(0)                              # shape: (C, L)

        # Extract motif tensors once from the original sequence
        motifs = [ohe[:, a:b].unsqueeze(0) for a, b in zip(starts, ends)]

        # For each shuffled background, substitute all motifs with given spacings
        dishuf_list = []
        for j in range(bgs.shape[0]):
            dishuf = multisubstitute(bgs[j].unsqueeze(0), motifs, spacing=spacings, start=int(starts[0]))
            # Ensure consistent shape (C, L)
            dishuf = dishuf.squeeze(0)
            dishuf_list.append(dishuf)

        # Stack into a tensor of shape (10, C, L) and store per sequence id
        shuffledbgs_byseq[bin_name][int(v)] = torch.stack(dishuf_list, dim=0)

print(f"\nTotal dishuffled backgrounds: {sum(sum(x.shape[0] for x in shuffledbgs_byseq[b].values()) for b in ['high', 'mid', 'low'])}")
print("Finished making all dishuffled bgs")
```

```python
import os, sys
import pytorch_lightning
model_dir  = "/grid/wsbs/home_norepl/pmantill/Trained_Model_Zoo/EvoAug_Distilled_Student_Model"
model_path = os.path.join(model_dir, "EvoAug_student_model.pt")

# Make sure distill_EvoAug2.py is in model_dir
sys.path.insert(0, model_dir)
import distill_EvoAug2 as distill_mod  # avoid shadowing 'model'

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
pytorch_model = torch.load(model_path, map_location=device).to(device).eval()

# Wrap to expose only the mean head (B,2)
class MeanOnlyModel(nn.Module):
    def __init__(self, student_model): 
        super().__init__()
        self.student = student_model
    def forward(self, X):
        out = self.student(X)   # dict
        return out["mean"]      # (B,2)

def uniquify_relu_calls(model: nn.Module) -> nn.Module:
    gm = fx.symbolic_trace(model)
    modules = dict(gm.named_modules())
    counter = 0
    for node in gm.graph.nodes:
        if node.op == 'call_module' and isinstance(modules[node.target], nn.ReLU):
            new_name = f"_relu_callsite_{counter}"
            counter += 1
            setattr(gm, new_name, nn.ReLU(inplace=False))
            node.target = new_name
    gm.recompile()
    return gm.eval()

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")  # Will use GPU 3 due to CUDA_VISIBLE_DEVICES
pytorch_model.to(device).eval()

attr_model = MeanOnlyModel(pytorch_model).eval()
attr_model = copy.deepcopy(attr_model).eval()
attr_model = uniquify_relu_calls(attr_model).to(device).eval()

torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True


def deeplift_attr_with_progress(x_np_BLA, target_idx, batch_size, n_shuffles):
    B, L, A = x_np_BLA.shape
    out = np.empty((B, L, A), dtype=np.float16)
    for s in range(0, B, batch_size):
        e = min(s + batch_size, B)
        x_batch = x_np_BLA[s:e]  # (b, L, A)
        x_t = torch.from_numpy(np.ascontiguousarray(x_batch)).to(device, dtype=torch.float32)
        x_t = x_t.permute(0, 2, 1).contiguous()          # -> (b, A, L)
        x_t.requires_grad_(True)                         # CRITICAL
        #print(target_idx)
        # IMPORTANT: do NOT wrap this in no_grad/inference_mode
        
        attr_BAL = deep_lift_shap(
            attr_model,
            x_t,
            target=target_idx,                # 0=Dev, 1=HK
            device='cuda',
            batch_size=x_t.shape[0],
            n_shuffles=n_shuffles,
            print_convergence_deltas=False,
            additional_nonlinear_ops={nn.ReLU: _nonlinear},
            verbose=True
        )

        attr_BLA = attr_BAL.permute(0, 2, 1).contiguous().cpu().detach().numpy()  # -> (b, L, A)
        out[s:e] = attr_BLA.astype(np.float16)

    return out
```

```python
tensor_list = [torch.tensor(arr) for arr in high_ohe_list]

# Get WT predictions
high_ohe_stack = torch.stack(tensor_list)
print(high_ohe_stack.shape)
y = predict(pytorch_model, high_ohe_stack, batch_size=256, func=lambda x: x['mean'])
y = y[:, 0]

scores_df = pd.DataFrame({"Test_idx": high_df['test_idx'], "WT_score":y})

## Collect WT scores for all bins
n_seqs = 100
wt_scores = {}  # wt_scores[bin] = list of scores

for bin_name in ['high', 'mid', 'low']:
    df = bin_data[bin_name]['df']
    ohe_list = bin_data[bin_name]['ohe']
    first = valid_seqs[bin_name][0:n_seqs]
    int_indices = list(map(int, first))
    
    # Get predictions for this bin's sequences
    tensor_list = [ohe_list[idx] for idx in int_indices]
    ohe_stack = torch.stack(tensor_list)
    y = predict(pytorch_model, ohe_stack, batch_size=256, func=lambda x: x['mean'])
    y = y[:, 0]
    
    wt_scores[bin_name] = y.tolist()
    
print(f"WT scores collected: {sum(len(wt_scores[b]) for b in ['high', 'mid', 'low'])} sequences")

## Collect dishuffled background predictions for all bins
# shuffledbgs_byseq[bin][seq_idx] has shape (10, C, L)
# NEW: Average the 10 shuffles per sequence

dishuff_scores = {}  # dishuff_scores[bin] = list of averaged scores (one per sequence)

for bin_name in ['high', 'mid', 'low']:
    avg_scores = []
    
    # Process each sequence separately to compute average of its 10 shuffles
    for seq_idx, T in shuffledbgs_byseq[bin_name].items():
        # T has shape (10, C, L)
        # Predict on all 10 shuffles
        y = predict(pytorch_model, T, batch_size=256, func=lambda x: x['mean'])
        y = y[:, 0]  # shape: (10,)
        
        # Average the 10 predictions
        avg_score = y.mean().item()
        avg_scores.append(avg_score)
    
    dishuff_scores[bin_name] = avg_scores
    
print(f"Dishuffled scores collected (averaged): {sum(len(dishuff_scores[b]) for b in ['high', 'mid', 'low'])} predictions")

## Collect swap predictions for all source->target bin combinations
# NEW: Average the 10 shuffles per (source_seq, target_seq) pair
n_seqs = 100
# swap_scores[source_bin][target_bin] = list of averaged predictions
swap_scores = {}

for source_bin in ['high', 'mid', 'low']:
    swap_scores[source_bin] = {}
    first_source = valid_seqs[source_bin][0:n_seqs]
    
    for target_bin in ['high', 'mid', 'low']:
        all_avg_preds = []
        
        for v in first_source:
            if v not in swaps[source_bin]:
                continue
            if target_bin not in swaps[source_bin][v]:
                continue
                
            for vt, ten in swaps[source_bin][v][target_bin].items():
                if len(ten) == 0:
                    continue
                # ten is a list of 10 swapped sequences
                input_tensor = torch.cat(ten, dim=0)  # (10,4,L)
                y = predict(pytorch_model, input_tensor, batch_size=256, func=lambda x: x['mean'])[:, 0]
                # y has shape (10,) - average these 10 shuffles
                avg_pred = y.mean().item()
                all_avg_preds.append(avg_pred)
        
        swap_scores[source_bin][target_bin] = all_avg_preds

print(f"Swap scores collected (averaged):")
for src in ['high', 'mid', 'low']:
    for tgt in ['high', 'mid', 'low']:
        print(f"  {src}->{tgt}: {len(swap_scores[src][tgt])} predictions")
```
